```python
from PIL import Image
import matplotlib.pyplot as plt

# Abrir a imagem
image = Image.open("/cat with glasses).jpeg")  # Substitua pelo nome do arquivo

# Converter para tons de cinza manualmente
pixels = image.load()  # Acessar os pixels da imagem
width, height = image.size

gray_image = Image.new("L", (width, height))  # Criar uma nova imagem em escala de cinza
binary_image = Image.new("L", (width, height))  # Criar uma nova imagem para binarização

threshold = 127  # Definir o limiar para binarização

for x in range(width):
    for y in range(height):
        r, g, b = pixels[x, y]  # Obter valores RGB
        gray_value = int(0.299 * r + 0.587 * g + 0.114 * b)  # Converter para escala de cinza

        gray_image.putpixel((x, y), gray_value)  # Atualizar pixel na imagem em tons de cinza

        # Aplicar binarização
        binary_value = 255 if gray_value > threshold else 0
        binary_image.putpixel((x, y), binary_value)

# Exibir as imagens no Colab
plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Imagem Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Imagem em Tons de Cinza")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(binary_image, cmap="gray")
plt.title("Imagem Binarizada")
plt.axis("off")

plt.show()
```
